```python
import pandas as pd
import numpy as np

df = pd.read_csv("anime_dataset.csv")
df
```

```python
# pare it down to just the columns we want
df = df[['title_english', 'synopsis', 'genres']]
df
```

```python
# get rid of rows with no synopsis or no genre
# df_clean = df.dropna(subset=['column_A', 'column_B'])
df = df.dropna(subset = ['synopsis', 'genres'])
df
```

```python
df = df[~df["synopsis"].str.startswith("Music video")]
df
```

```python
df = df.drop_duplicates(subset = "title_english")
df = df.drop_duplicates(subset = "synopsis")
df
```

```python
# replace missing titles with... should I AI-generate new titles? or google what the titles are supposed to be 
# df[df['Age'] > 30]
# ok after cleaning there is only one missing title. so maybe I'll just get rid of it lol 
df_missingtitle = df[df['title_english'].isna()]
df_missingtitle

df = df[~df["synopsis"].str.startswith('Shy, reserved, and small-statured, Deimon High')]
df
```

```python
# japanese check: I couldn't find any after looking at rows that were flagged for japanese, so 
# i'm going to assume I got all the japanese out by taking out duplicates. 

# \u3040-\u309f: Hiragana | \u30a0-\u30ff: Katakana | \u4e00-\u9faf: Kanji
japanese_pattern = r"[\u3040-\u309f\u30a0-\u30ff\u4e00-\u9faf]"

# 2. Check if any cell matches the pattern
# (We cast to string first to avoid errors with numbers or NaNs)
has_japanese = (
    df.map(lambda x: bool(pd.Series(str(x)).str.contains(japanese_pattern).iloc[0]))
    .any()
    .any()
)

print(f"Contains Japanese: {has_japanese}")

rows_with_japanese = df[
    df.map(lambda x: bool(pd.Series(str(x)).str.contains(japanese_pattern).iloc[0])).any(
        axis=1
    )
]
print(rows_with_japanese)
```

```python
# feels done to me. let's write to csv and look 
# reset index
df = df.reset_index()
df = df[['title_english', 'synopsis', 'genres']]
df.to_csv("cleaned_animes.csv")
```
